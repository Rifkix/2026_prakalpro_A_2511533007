## 1.1 Latar Belakang
Percabangan (branching) merupakan salah satu struktur kontrol dasar dalam pemrograman yang memungkinkan suatu program mengambil keputusan dan menjalankan blok kode yang berbeda berdasarkan kondisi tertentu. Tanpa percabangan, sebuah program hanya dapat berjalan secara linear dari baris pertama hingga terakhir tanpa mampu merespons variasi input yang diberikan oleh pengguna. Dalam bahasa Python, percabangan diimplementasikan melalui beberapa bentuk, yaitu *if* tunggal, *if-else*, *if-elif-else* yang dapat dikombinasikan dengan operator logika (*and*, *or*, *!=*), *multi-if* terpisah untuk kondisi yang dievaluasi secara independen, serta *match-case* sebagai struktur pencocokan pola yang setara dengan *switch-case* pada bahasa lain.

## 2.1 Dasar Teori
Dasar teori dalam praktikum ini meliputi lima bentuk struktur percabangan dalam Python:

1. **If Tunggal**: Struktur percabangan paling sederhana yang hanya mengeksekusi satu blok kode apabila kondisinya bernilai *True*, dan tidak melakukan apa pun jika kondisinya *False*. Bentuk ini umumnya digunakan untuk validasi sederhana, seperti memeriksa apakah suatu input logis atau tidak.

2. **If-Else**: Digunakan ketika terdapat dua kemungkinan alur eksekusi. Jika kondisi bernilai *True* maka blok *if* dijalankan, sebaliknya jika *False* maka blok *else* yang dijalankan.

3. **If-Elif-Else dengan Operator Logika**: Digunakan untuk mengevaluasi beberapa kondisi secara berurutan, di mana hanya satu blok yang akan dieksekusi sesuai kondisi pertama yang terpenuhi. Operator logika seperti *and* (kedua kondisi harus benar), *or* (salah satu kondisi benar), dan *!=* (tidak sama dengan) digunakan untuk membentuk kondisi majemuk, misalnya menggabungkan syarat umur dan kepemilikan SIM sekaligus.

4. **Multi-IF Terpisah**: Berbeda dengan *if-elif-else* yang hanya mengeksekusi satu cabang, *multi-if* terpisah menggunakan beberapa blok *if* independen yang dievaluasi satu per satu tanpa saling bergantung. Struktur ini memungkinkan beberapa kondisi terpenuhi secara bersamaan sehingga hasilnya dapat diakumulasikan, seperti pada perhitungan diskon bertingkat (diskon belanja besar, diskon member, diskon promo, dan diskon rombongan yang dapat ditumpuk).

5. **Match-Case**: Struktur pencocokan pola yang diperkenalkan pada Python 3.10, setara dengan *switch-case* pada bahasa pemrograman lain. Setiap *case* mencocokkan nilai variabel dengan pola tertentu, dan *case _* berfungsi sebagai kondisi *default* yang menangani nilai di luar pola yang telah didefinisikan. Struktur ini lebih ringkas dibandingkan rangkaian *if-elif* panjang ketika pilihan bersifat diskret, seperti pemilihan paket wahana bernomor 1 sampai 5.

### 2.2.1 Program "if1_2511533007"

```python
#Program ini menggunakan fungsi input()

ipk_3007 = float(input("Input IPK Anda = "))

if ipk_3007 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3007))

print("Program Selesai")
```

```python
ipk_3007 = float(input("Input IPK Anda = "))

if ipk_3007 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3007))
else:
    print("Anda Tidak lulus")
print("Program Selesai")
```

**Penjelasan :**

### 2.2.3 Program "multi_if1_2511533007"

```python
umur_3007= int(input("Input umur anda: "))
sim_3007 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0]

if umur_3007 >= 17 and sim_3007 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3007 >= 17 and sim_3007 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3007 < 17 and sim_3007 == 'y':
    print("Anda Belum Cukup umur punya SIM")

if umur_3007 < 17 and sim_3007 != 'y':
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan :**

2. program akan menjalankan if jika kondisi if terpenuhi,
contohnya kode `if umur_3007 >= 17 and sim_3007 == 'y':` dijalankan jika inputan umur besar atau sama dengan 17 dan inputan sim adalah y

### 2.2.4 Program "multi_if2_2511533007"

```python
#Program Menghitung Diskon Belanja

#input dari user
total_belanja_3007 = float(input("Masukkan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3007 = input("Apakah Anda member(y/t): ").strip().lower()
is_member_3007 = input_member_3007 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3007 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3007 = input_promo_3007 in ["y", "ya"]

total_diskon_persen_3007 = 0

# Multi IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3007 > 1000000:
    total_diskon_persen_3007 += 10 #diskon belanja besar

if is_member_3007:
    total_diskon_persen_3007 += 5 # diskon member

if kode_promo_valid_3007:
    total_diskon_persen_3007 += 15 # diskon voucher

# Menghitung nominal diskon dan total baya
nominal_diskon_3007 = total_belanja_3007 * (total_diskon_persen_3007/100)
total_bayar_3007 = total_belanja_3007 - nominal_diskon_3007

#Ouput hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total diskon : {total_diskon_persen_3007}% (Rp {nominal_diskon_3007:,.0f})")
print(f"Total bayar : Rp {total_bayar_3007:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3007}%")
#ouput: total diskon yang anda dapatkan: 30 jika belanja > 1 juta, member dan kode promo valid
```

**Penjelasan :**
1. Program membaca inputan user berupa total belanja, apakah member, dan apakah punya kode promo
2. Jika total belanja besar dari 1 juta maka diskon + 10%, jika member maka diskon + 5%, jika punya kode promo maka diskon + 15%
3. kemudian total bayar yang ada itu dikurangi nominal diskon

### 2.2.5 Program "if_elif_else1_2511533007"

```python
umur_3007 = int(input("Input umur anda: "))
sim_3007 = input("Apakah anda sudah punya SIM C: ")[0]

if umur_3007 >= 17 and sim_3007 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

elif umur_3007 >= 17 and sim_3007 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

elif umur_3007 < 17 and sim_3007 == 'y':
    print("Anda Belum Cukup umur punya SIM")

else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program selesai")
```

**Penjelasan :**
1. bedanya if, if, if, ... dengan if elif else adalah: untuk if, if, if, ..., jika if pertama benar dan if kedua sampai seterusnya benar, maka program akan menjalankan semua if yang benar, sedangkan if elif else akan memilih siapa yang benar duluan dari atas, jika ada yang sudah benar, maka elif dan else dibawahnya tidak akan dijalankan, jika if dan elif tidak ada yang bernilai benar, maka else akan dijalankan 

### 2.2.6 Program "match_case_2511533007"

```python
#Program konversi angka menjadi nama bulan

bulan_3007 = int(input("Masukkan angka bulan (1-12): "))

match bulan_3007:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
        
```

**Penjelasan :**
1. program meminta inputan angka, jika angka 1 yang di inputkan, maka akan menghasilkan January, begitu seterusnya sampe inputan angka 12, tapi jika user tidak menginputkan int dengan rentang 1-12 maka akan menjalankan case _ dan memprint Angka tidak valid

### 2.2.7 Program "nilai_2511533007"

```python
#program konversi nilai angka menjadi huruf

nilai_3007 = int(input("Inputkan nilai angka = "))

if nilai_3007 >= 81:
    print("A")
elif nilai_3007 >= 70:
    print("B")
elif nilai_3007 >= 60:
    print("C")
elif nilai_3007 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan:** 
1. program meminta inputan integer, jika nilai nya besar sama dengan 81 maka A, begitu seterusnya hinga jika inputan besar sama dengan 50 maka D, dan jika semuanya tidak memenuhi barulah menjalankan else yaitu nilai E

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")


# 1. Input Data Pengunjung & String Handling
nama_3007 = input("Masukkan Nama Pengunjung        : ")
umur_3007 = int(input("Input umur anda                 : "))
sim_3007 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

# 2. Pemilihan Wahana menggunakan match-case
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_3007 = int(input("Masukkan nomor paket (1-5)      : "))

nama_wahana_3007 = ""
harga_satuan_3007 = 0

match paket_3007:
    case 1:
        nama_wahana_3007 = "Wahana Safari Rimba"
        harga_satuan_3007 = 50000
    case 2:
        nama_wahana_3007 = "Wahana Arung Jeram"
        harga_satuan_3007 = 75000
    case 3:
        nama_wahana_3007 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3007 = 120000
    case 4:
        nama_wahana_3007 = "Wahana Roller Coaster Kilat"
        harga_satuan_3007 = 100000
    case 5:
        nama_wahana_3007 = "Wahana All-Access VIP"
        harga_satuan_3007 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

jumlah_tiket_3007 = int(input("Masukkan jumlah tiket           : "))

if jumlah_tiket_3007 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

input_member_3007 = input("Apakah Anda member? (y/t)       : ").strip().lower()
is_member_3007 = input_member_3007 in ["y", "ya"]

input_promo_3007 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()
kode_promo_valid_3007 = input_promo_3007 in ["y", "ya"]


# 3. Validasi Izin Kendali Wahana - if-elif-else + operator logika
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_3007 == 3 and umur_3007 >= 17 and sim_3007 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif paket_3007 == 3 and umur_3007 >= 17 and sim_3007 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif paket_3007 == 3 and umur_3007 < 17 and sim_3007 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif paket_3007 == 3 and umur_3007 < 17 and sim_3007 != 'y':
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif paket_3007 != 3 and umur_3007 >= 10:
    print("Status Akses: Anda memenuhi syarat umur untuk menikmati wahana ini.")
else:
    print("Status Akses: Anda belum cukup umur untuk menikmati wahana ini.")


# 4. Akumulasi Diskon Bertingkat - Multi-IF Terpisah
subtotal_3007 = harga_satuan_3007 * jumlah_tiket_3007
total_diskon_persen_3007 = 0

if subtotal_3007 >= 200000:
    total_diskon_persen_3007 += 10  # Diskon Belanja Besar

if is_member_3007:
    total_diskon_persen_3007 += 5  # Diskon Member

if kode_promo_valid_3007:
    total_diskon_persen_3007 += 15  # Diskon Voucher Promo

if jumlah_tiket_3007 >= 5:
    total_diskon_persen_3007 += 5  # Diskon Tambahan Rombongan

nominal_diskon_3007 = subtotal_3007 * (total_diskon_persen_3007 / 100)
total_bayar_3007 = subtotal_3007 - nominal_diskon_3007


# 5. Evaluasi Kelulusan Audit - if-else
if total_bayar_3007 > 300000:
    catatan_layanan_3007 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_3007 = "Terima kasih telah berkunjung."


print("\n--- Rincian Pembayaran ---")
print(f"Nama Pengunjung  : {nama_3007}")
print(f"Wahana Dipilih   : {nama_wahana_3007}")
print(f"Jumlah Tiket     : {jumlah_tiket_3007}")
print(f"Subtotal Belanja : Rp {subtotal_3007:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_3007}% (Rp {nominal_diskon_3007:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_3007:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_3007}")
print("Program Selesai")
```

**Penjelasan :** 
Pertama, program mengambil data pengunjung (nama, umur, status SIM C) lalu menampilkan daftar lima paket wahana dan meminta pengguna memilih nomor paket — pemilihan ini ditangani oleh match-case, yang menentukan nama_wahana_3007 dan harga_satuan_3007 sesuai case yang cocok, atau menghentikan program (exit()) jika input di luar 1–5. Setelah itu, jumlah tiket diminta dan langsung dicek dengan if tunggal: jika ≤ 0, program hanya mencetak peringatan tanpa menghentikan eksekusi. Status member dan validitas kode promo juga dikonversi menjadi boolean (True/False) di tahap ini.

Selanjutnya program mengevaluasi kelayakan pengendara melalui satu rangkaian if-elif-else. Karena aturan aslinya berbeda antara paket 3 (ATV) dan paket lainnya, setiap cabang kondisi digabungkan dengan syarat paket_3007 == 3 atau paket_3007 != 3 menggunakan operator and — sehingga tetap satu chain datar (bukan if di dalam if) namun tetap bisa membedakan keempat kombinasi umur/SIM untuk ATV serta aturan umur ≥ 10 untuk paket lain.

Bagian terakhir menghitung subtotal_3007 (harga × jumlah tiket), lalu memeriksa empat syarat diskon lewat empat blok if yang terpisah (bukan elif), sehingga persentase diskon bisa terakumulasi jika lebih dari satu syarat terpenuhi sekaligus.

## 2.1 Dasar Teori
Dasar teori dalam praktikum ini meliputi beberapa konsep dasar pemrograman Python:

1. **Operator Aritmatika**: Digunakan untuk operasi matematika dasar seperti penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), pangkat (**), dan modulus/sisa bagi (%).

2. **Operator Perbandingan**: Digunakan untuk membandingkan dua buah nilai. Operator ini selalu mengembalikan nilai boolean True atau False (contoh: ==, !=, >, <, >=, <=).

3. **Operator Logika**: Digunakan untuk menggabungkan pernyataan bersyarat atau nilai boolean (yaitu and, or, dan not).

4. **Operator Penugasan (Assignment)**: Digunakan untuk memberikan nilai ke dalam suatu variabel. Python mendukung augmented assignment di mana penugasan digabung dengan operasi aritmatika (contoh: +=, -=, *=).

5. **Operator Keanggotaan (Membership)**: Digunakan untuk memeriksa apakah sebuah elemen atau nilai tertentu terdapat di dalam sebuah sequence seperti list, string, atau tuple (in dan not in).

6. **Operator Identitas (Identity)**: Digunakan untuk membandingkan objek di dalam memori, bukan sekadar membandingkan kesamaan nilainya (is dan is not).

7. **Operator Bitwise**: Digunakan untuk membandingkan angka (integer) pada tingkat bit atau biner (contoh: AND &, OR |, XOR ^, NOT ~, Shift Kiri <<, Shift Kanan >>).

## 2.2 Kode Program

### 2.2.1 Program "Aritmatika_2511533007"

```python
#program ini menggunakan fungsi input()
#nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3007 = int(input("Input angka-1: "))
angka2_3007 = int(input("Input angka-2: "))

#penjumlahan
hasil_3007 = angka1_3007 + angka2_3007
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3007)

#pengurangan
hasil_3007 = angka1_3007 - angka2_3007
print("\nOperator Pengurangan")
print("Hasil =", hasil_3007)

#perkalian
hasil_3007 = angka1_3007 * angka2_3007
print("\nOperator Perkalian")
print("Hasil =", hasil_3007)

#pembagian, pembagian bulat, dan sisa bagi
if angka1_3007 != 0:
    hasil_3007 = angka1_3007 / angka2_3007
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3007)

    hasil_3007 = angka1_3007 // angka2_3007
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3007)

    hasil_3007 = angka1_3007 % angka2_3007
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_3007)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# pangkat
hasil_3007 = angka1_3007 ** angka2_3007
print("\nOperator pangkat")
print("Hasil =", hasil_3007)
```

**Penjelasan :**
1. Program meminta dua angka dari pengguna dan mengubahnya menjadi integer.
2. Program menghitung penjumlahan, pengurangan, perkalian, pembagian, pembagian bulat, sisa bagi, dan perpangkatan.
3. Hasil setiap operasi ditampilkan ke layar.

### 2.2.2 Program "Assignment_2511533007"

```python
# Program operator assignment dalam python

angka1_3007 = int(input("Input angka-1: "))
angka2_3007 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_3007)
print("Nilai angka2 =", angka2_3007)

# assignment biasa
hasil_3007 = angka1_3007
print("\nAssignment Biasa (=)")
print("Hasil =", hasil_3007)

# assignment penambahan
hasil_3007 = angka1_3007
hasil_3007 += angka2_3007
print("\nAssignment Penambahan (+=)")
print("Hasil =", hasil_3007)

# assignment pengurangan
hasil_3007 = angka1_3007
hasil_3007 -= angka2_3007
print("\nAssignment Pengurangan (-=)")
print("Hasil =", hasil_3007)

# assignment perkalian
hasil_3007 = angka1_3007
hasil_3007 *= angka2_3007
print("\nAssignment Perkalian (*=)")
print("Hasil =", hasil_3007)

# assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3007 != 0:
    hasil_3007 = angka1_3007
    hasil_3007 /= angka2_3007
    print("\nAssignment Pembagian (/=)")
    print("Hasil =", hasil_3007)
    #Operator tambahan
    hasil_3007 = angka1_3007
    hasil_3007 //= angka2_3007
    print("\nAssignment Pembagian Bulat (//=)")
    print("Hasil =", hasil_3007)
    hasil_3007 = angka1_3007
    hasil_3007 %= angka2_3007
    print("\nAssignment Sisa Bagi (%=)")
    print("Hasil =", hasil_3007)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1_3007
hasil **= angka2_3007
print("\nAssignment Perpangkatan (**=)")
print("Hasil =", hasil)
```

**Penjelasan :**
1. Program membaca dua angka dari pengguna.
2. Program menerapkan operator assignment seperti =, +=, -=, *=, /=, //=, %=, dan **=.
3. Setiap hasil operasi assignment ditampilkan untuk menunjukkan perubahan nilai.

### 2.2.3 Program "Perbandingan_2511533007"

```python
#program operator perbandingan dalam python

angka1_3007 = int(input("Input angka-1: "))
angka2_3007 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_3007 = angka1_3007 > angka2_3007
print("\nOperator Lebih Besar Dari")
print("angka1 > angka2 =", hasil_3007)

# Lebih kecil dari
hasil_3007 = angka1_3007 < angka2_3007
print("\nOperator Lebih Kecil Dari")
print("angka1 < angka2 =", hasil_3007)

# Lebih besar dari atau sama dengan
hasil_3007 = angka1_3007 >= angka2_3007  
print("\nOperator Lebih Besar Dari atau Sama Dengan")
print("angka1 >= angka2 =", hasil_3007)

# Lebih kecil dari atau sama dengan
hasil_3007 = angka1_3007 <= angka2_3007
print("\nOperator Lebih Kecil Dari atau Sama Dengan")
print("angka1 <= angka2 =", hasil_3007)

# Sama dengan
hasil_3007 = angka1_3007 == angka2_3007
print("\nOperator Sama Dengan")
print("angka1 == angka2 =", hasil_3007)

# Tidak sama dengan
hasil_3007 = angka1_3007 != angka2_3007
print("\nOperator Tidak Sama Dengan")
print("angka1 != angka2 =", hasil_3007)

# tambahan: perbandingan berantai dalam python
hasil_3007 = 0 < angka1_3007 < 100
print("\nPerbandingan Berantai")
print("0 < angka1 < 100 =", hasil_3007)

hasil_3007 = 0 < angka2_3007 < 100
print("0 < angka2 < 100 =", hasil_3007)
```

**Penjelasan :**
1. Program meminta dua angka dan membandingkannya menggunakan operator perbandingan.
2. Output yang dihasilkan berupa nilai boolean True atau False.
3. Program juga menampilkan contoh perbandingan berantai yang khas dimiliki Python.

### 2.2.4 Program "Logika_2511533007"

```python
# prorgam operator logika dalam python

# Memasukkan nilai boolean
# input tidak peka terhadap huruf besar dan kecil
a1_3007 = input("Input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_3007 = input("Input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 =", a1_3007)
print("A2 =", a2_3007)

# Konjungsi : bernilai True jika keduanya True
hasil_3007 = a1_3007 and a2_3007
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_3007)

# Disjungsi : bernilai True jika salah satu True
hasil_3007 = a1_3007 or a2_3007
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_3007)

# Negasi A1: membalik nilai A1
hasil_3007 = not a1_3007
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_3007)

# Negasi A2: membalik nilai A2
hasil_3007 = not a2_3007
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_3007)

# XOR bernilai True jika kedua nilai berbeda
hasil_3007 = a1_3007 != a2_3007
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_3007)
```

**Penjelasan :**
1. Program menerima dua nilai boolean dari pengguna.
2. Program menghitung operasi logika AND, OR, NOT, dan XOR.
3. Hasil operasi logika ditampilkan dalam bentuk boolean.

### 2.2.5 Program "Lainnya_2511533007"

```python
#Program operator keanggotaan dan identitas

print("=====================================")
print("1. OPERATOR KEANGGOTAAN")
print("=====================================")

#input beberapa data yang dipisahkan dengan koma
input_data_3007 = input("Masukkan beberapa angka, pisahkan dengan koma:")

#mengubah input menjadi list integer
data_3007 = [int(angka_3007.strip()) for angka_3007 in input_data_3007.split(",")]

nilai_dicari_3007 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3007 = nilai_dicari_3007 in data_3007
print("\nOperator keanggotaan IN")
print("Apakah", nilai_dicari_3007, "in", data_3007, "=", hasil_3007)

#operator not in
hasil_3007 = nilai_dicari_3007 not in data_3007
print("\nOperator keanggotaan NOT IN")
print("Apakah", nilai_dicari_3007, "not in", data_3007, "=", hasil_3007)

print("\n=====================================")
print("2. OPERATOR IDENTITAS")
print("=====================================")

# objek1 menggunakan list dari input pengguna
objek1_3007 =data_3007

# objek2 merujuk pada objek yang sama dengan objek1
objek2_3007 = objek1_3007

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3007 = data_3007.copy()

print("objek1 =", objek1_3007)
print("objek2 =", objek2_3007)
print("objek3 =", objek3_3007)

# Operator is
hasil_3007 = objek1_3007 is objek2_3007
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_3007)

# Operator is not
hasil_3007 = objek1_3007 is not objek3_3007
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_3007)

# membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 == objek3 =", objek1_3007 == objek3_3007)
print("objek1 == objek3 =", objek1_3007 is objek3_3007)
```

**Penjelasan :**
1. Program meminta beberapa angka yang dipisahkan koma, lalu mengubahnya menjadi list integer.
2. Program memeriksa apakah suatu angka terdapat dalam list menggunakan in dan not in.
3. Program membandingkan identitas objek menggunakan is dan is not, serta membandingkan nilai dengan ==.

### 2.2.6 Program "Bitwise_2511533007"

```python
print("\n=====================================")
print("3. OPERATOR BITWISE")
print("=====================================")

angka1_3007 = int(input("Input angka bitwise-1: "))
angka2_3007 = int(input("Input angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_3007, "| biner =", bin(angka1_3007))
print("angka2 =", angka2_3007, "| biner =", bin(angka2_3007))

# Bitwise AND
hasil_3007 = angka1_3007 & angka2_3007
print("\nOperator Bitwise AND (&)")
print(angka1_3007, "&", angka2_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

# Bitwise OR
hasil_3007 = angka1_3007 | angka2_3007
print("\nBitwise OR (|)")
print(angka1_3007, "|", angka2_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

# Bitwise XOR
hasil_3007 = angka1_3007 ^ angka2_3007
print("\nBitwise XOR (^)")
print(angka1_3007, "^", angka2_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

# Bitwise NOT
hasil_3007 = ~angka1_3007
print("\nBitwise NOT (~)")
print("~", angka1_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

# Bitwise geser kiri
jumlah_geser_3007 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3007 = angka1_3007 << jumlah_geser_3007
print("\nBitwise Geser Kiri (<<)")
print(angka1_3007, "<<", jumlah_geser_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

# Bitwise geser kanan
hasil_3007 = angka1_3007 >> jumlah_geser_3007
print("\nBitwise Geser Kanan (>>)")
print(angka1_3007, ">>", jumlah_geser_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))
```

**Penjelasan :**
1. Program menerima dua angka untuk operasi bitwise.
2. Program menampilkan bentuk biner dari angka yang dimasukkan.
3. Program menghitung operasi AND, OR, XOR, NOT, geser kiri, dan geser kanan, lalu menampilkan hasilnya dalam desimal dan biner.

```python
# === SISTEM SIMULASI TRANSAKSI DAN VALIDASI AKSES TOKO ===

# 1. INPUT DATA (Menerima input dari pengguna)
print("=== SISTEM TRANSAKSI TOKO ===\n")
nama_3007 = input("Masukkan Nama Pelanggan : ")
status_3007 = input("Masukkan Status Pelanggan (member/nonmember) : ").lower()
total_belanja_3007 = int(input("Masukkan Total Belanja : "))
jumlah_barang_3007 = int(input("Masukkan Jumlah Barang : "))
kode_promo_3007 = input("Masukkan Kode Promo : ").upper()

print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan       : {nama_3007}")
print(f"Status Pelanggan     : {status_3007}")
print(f"Total Belanja        : Rp{total_belanja_3007}")
print(f"Jumlah Barang        : {jumlah_barang_3007}")
print(f"Kode Promo           : {kode_promo_3007}")

# 2. OPERATOR PERBANDINGAN & KEANGGOTAAN
daftar_promo_3007 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]
cek_belanja_3007 = total_belanja_3007 >= 200000    # Operator >=
cek_barang_3007 = jumlah_barang_3007 >= 3          # Operator >=
cek_member_3007 = status_3007 == "member"          # Operator ==
promo_tersedia_3007 = kode_promo_3007 in daftar_promo_3007 # Operator IN

# 3. OPERATOR LOGIKA
# Diskon didapat jika member AND belanja minimum tercapai
dapat_diskon_3007 = cek_member_3007 and cek_belanja_3007
# Promo didapat jika kode valid AND (barang cukup OR sudah dapat diskon)
dapat_promo_3007 = promo_tersedia_3007 and (cek_barang_3007 or dapat_diskon_3007)
# Contoh penggunaan operator not
bukan_member_3007 = not cek_member_3007

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {cek_belanja_3007}")
print(f"Jumlah Barang >= 3         : {cek_barang_3007}")
print(f"Status Member              : {cek_member_3007}")
print(f"Kode Promo Tersedia        : {promo_tersedia_3007}")
print(f"Mendapatkan Diskon         : {dapat_diskon_3007}")
print(f"Mendapatkan Promo          : {dapat_promo_3007}")

# 4. OPERATOR ARITMATIKA & PENUGASAN (AUGMENTED ASSIGNMENT)
diskon_3007 = 0
if dapat_diskon_3007:
    diskon_3007 = total_belanja_3007 * 0.10  # Operator * (Perkalian)

total_bayar_3007 = total_belanja_3007
total_bayar_3007 -= diskon_3007              # Operator Penugasan -=

# Operator / (Pembagian) dan % (Modulus/Sisa Bagi)
rata_rata_3007 = total_bayar_3007 / jumlah_barang_3007 if jumlah_barang_3007 > 0 else 0
sisa_bagi_3007 = int(total_bayar_3007) % jumlah_barang_3007 

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                     : Rp{int(diskon_3007)}")
print(f"Total Pembayaran           : Rp{int(total_bayar_3007)}")
print(f"Rata-rata Harga Barang     : Rp{int(rata_rata_3007)}")

# 5. OPERATOR BITWISE
# Representasi bit: Member(0001), Belanja(0010), Barang(0100), Promo(1000)
bit_member_3007 = 1 if cek_member_3007 else 0       
bit_belanja_3007 = 2 if cek_belanja_3007 else 0     
bit_barang_3007 = 4 if cek_barang_3007 else 0       
bit_promo_3007 = 8 if promo_tersedia_3007 else 0    

# OR (|) untuk menggabungkan kondisi
status_transaksi_3007 = bit_member_3007 | bit_belanja_3007 | bit_barang_3007 | bit_promo_3007

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses             : {format(status_transaksi_3007, '04b')}")
# AND (&) untuk mengekstrak/memeriksa hak akses
print(f"Member Access              : {(status_transaksi_3007 & 1) > 0}")
print(f"Promo Access               : {(status_transaksi_3007 & 8) > 0}")
print(f"Free Shipping Access       : {(status_transaksi_3007 & 4) > 0}")

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print("0001 | 0010 | 0100 | 1000")
print(f"Kode Biner   : {format(status_transaksi_3007, '04b')}")
print(f"Kode Desimal : {status_transaksi_3007}")

print("=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{format(status_transaksi_3007, '04b')} & 0001")
cek_bit_member_3007 = status_transaksi_3007 & 1
print(f"Hasil Biner   : {format(cek_bit_member_3007, '04b')}")
print(f"Hasil Desimal : {cek_bit_member_3007}")

print("\nCek Promo")
print(f"{format(status_transaksi_3007, '04b')} & 1000")
cek_bit_promo_3007 = status_transaksi_3007 & 8
print(f"Hasil Biner   : {format(cek_bit_promo_3007, '04b')}")
print(f"Hasil Desimal : {cek_bit_promo_3007}")

print("=== Perbandingan Status ===")
# Misal membandingkan dengan profil target (1011 = Member, Belanja>=200k, Promo valid)
kode_referensi_3007 = 11 
print(f"Kode Transaksi : {format(status_transaksi_3007, '04b')}")
print(f"Kode Referensi : {format(kode_referensi_3007, '04b')}")
print(f"{format(status_transaksi_3007, '04b')} ^ {format(kode_referensi_3007, '04b')}")
hasil_xor_3007 = status_transaksi_3007 ^ kode_referensi_3007
print(f"Hasil Biner   : {format(hasil_xor_3007, '04b')}")
print(f"Hasil Desimal : {hasil_xor_3007}")

print("=== Shift ===")
print(f"{format(status_transaksi_3007, '04b')} << 1")
hasil_shift_3007 = status_transaksi_3007 << 1
print(f"Hasil Biner   : {format(hasil_shift_3007, '05b')}")
print(f"Hasil Desimal : {hasil_shift_3007}")

# 6. OPERATOR IDENTITAS (Tambahan Bukti Operasi)
print("\n=== DEMONSTRASI OPERATOR IDENTITAS ===")
objek1_3007 = daftar_promo_3007
objek2_3007 = objek1_3007
objek3_3007 = daftar_promo_3007.copy()

print(f"objek1 is objek2      = {objek1_3007 is objek2_3007}")
print(f"objek1 is not objek3  = {objek1_3007 is not objek3_3007}")
print(f"objek1 == objek3      = {objek1_3007 == objek3_3007} (Isi sama, namun identitas di memori berbeda)")

print("\n=== SELESAI ===")
```

Begitu sistem mengumpulkan seluruh masukan tersebut, alur berpindah pada fase verifikasi dasar yang menggunakan operator perbandingan dan keanggotaan. Di tahap ini, program membandingkan input dengan parameter standar, seperti apakah belanja melampaui Rp200.000, jumlah barang lebih dari tiga, hingga memastikan validitas kode promo yang diajukan pelanggan terhadap daftar promo (list) yang diizinkan. Pengecekan ini diolah untuk menghasilkan status True atau False yang kemudian saling dikombinasikan melalui operator logika. Proses logis ini bertugas menyaring kondisi gabungan—misalnya, sistem baru akan mengizinkan pemotongan harga diskon apabila profil yang diperiksa memegang status member (True) sekaligus total belanja memenuhi syarat minimum (True).


Setelah rangkaian validasi persetujuan selesai, program melaju ke proses perhitungan aritmatika dan manajemen hak akses menggunakan instrumen bitwise. Jika pelanggan terbukti memenuhi kriteria diskon berdasarkan alur logika sebelumnya, program otomatis mengaktifkan operator aritmatika, yaitu mengalikan (*) dasar harga dengan persentase tertentu, lalu mengurangi tagihan akhir tersebut memakai operator penugasan augmented (-=). Bersamaan dengan operasi finansial, sistem menyematkan kode akses melalui mekanisme manipulasi bit. Tiap syarat yang lolos dipetakan menjadi nilai bit biner spesifik (misal: 0001 untuk member, 1000 untuk kode promo), lalu digabungkan menjadi sebuah satu-kesatuan identitas numerik transaksi menggunakan operator Bitwise OR (|). Hasil dari penggabungan ini memungkinkan sistem untuk menguji kembali hak pengguna secara mandiri menggunakan operasi pengirisan bit, hingga keseluruhan ringkasan data, kelayakan, hasil hitung tunai, dan detail analisis sistem tercetak utuh di layar pengguna.

### Pertanyaan Analisis

1. Penjelasan Operator Aritmatika
<br>Perkalian (*): Digunakan untuk mengalikan total belanja dengan 0.10 dalam menentukan besaran nominal diskon 10%.  Pembagian (/): Digunakan untuk mencari harga rata-rata dari setiap barang dengan membagi total pembayaran terhadap jumlah barang.  Modulus (%): Digunakan untuk memperoleh sisa pembagian bulat dari perhitungan total pembayaran terhadap jumlah barang.

2. Perbedaan Operator and, or, dan not dalam Program
<br>and (Konjungsi): Menghasilkan True hanya jika seluruh kondisi terpenuhi. Dalam program, diskon hanya diberikan jika pelanggan berstatus member dan nominal belanjanya memenuhi syarat (keduanya wajib).  or (Disjungsi): Menghasilkan True jika salah satu dari kondisi terpenuhi. Pada program, promo dapat diaplikasikan jika jumlah barang >= 3 atau pelanggan telah terkualifikasi mendapatkan diskon (cukup salah satu).  not (Negasi): Berfungsi membalik nilai boolean yang dievaluasi. Dalam skenario program ini, not cek_member digunakan untuk menghasilkan True bila profil pelanggan merupakan nonmember.

3. Fungsi Operator in dan not in
<br>in: Melakukan pengecekan apakah nilai dari input pengguna (misal: "HEMAT10") terdapat di dalam list daftar_promo_3007.  not in: Merupakan kebalikannya; menghasilkan True jika input tidak direkam dalam list tersebut, sehingga dapat digunakan untuk memvalidasi penolakan promo tidak terdaftar.

4. Perbedaan is dan ==
<br>== (Equality): Mengevaluasi kesamaan nilai atau isi. Jika dua list memiliki data elemen yang persis sama, == akan menghasilkan True.  is (Identity): Mengevaluasi identitas objek (alamat memori). Meskipun dua list isinya sama persis, jika mereka dikerjakan dengan fungsi .copy() atau dideklarasikan terpisah, keduanya akan tersimpan di ruang memori yang berbeda sehingga menghasilkan output False saat menggunakan is.

5. Penggunaan Operator Bitwise untuk Hak Akses
<br>Dalam simulasi ini, hak akses pelanggan dimodelkan ke dalam representasi biner berdasarkan kombinasi syarat tertentu:Setiap syarat yang terpenuhi akan dikonversi menjadi bit independen (0001 untuk member, 1000 untuk promo, dll).Operator OR (|) menggabungkan bit-bit individu tersebut menjadi sebuah Kode Status Transaksi (misalnya 1001 berarti pelanggan adalah member dan memiliki akses promo).  Operator AND (&) kemudian dipakai sebagai masking filter. Misalnya untuk mengecek hak akses promo, program memeriksa status transaksi dengan & 1000. Jika hasilnya > 0, berarti hak akses promo diizinkan.  Operator XOR (^) dipakai untuk mencari letak ketidaksesuaian/perbedaan status antara status pelanggan riil dengan referensi standar toko.

6. Alur Eksekusi Program
<br>Input Data: Program menerima input tipe data campuran (nama, status, harga, barang, promo).Validasi Kelayakan: Menggunakan operator perbandingan dan keanggotaan untuk menentukan batasan minimum belanja, kuantitas, dan ketersediaan kupon.Proses Logika: Menggabungkan hasil perbandingan dengan logika boolean (and, or) untuk mengunci apakah pelanggan sah mendapatkan hak diskon/promo tambahan.Perhitungan Harga: Menghitung nominal menggunakan operator aritmatika (seperti diskon dari persentase) dan mengaplikasikan operator penugasan (-=) untuk mengurangi total tagihan dasar.  Pembentukan Kode Akses: Menerjemahkan setiap variabel kelayakan Boolean menjadi integer Bitwise, yang kemudian dicetak utuh mulai dari versi biner dan desimal hingga pengujian hak akses spesifik per kode biner.

7. Identifikasi Kondisi True dan False (Berdasarkan contoh input Anita)
<br>Kondisi yang menghasilkan True: cek_belanja_3007 bernilai True karena input total belanjanya adalah Rp250.000 (Syaratnya >= Rp200.000).
<br>Kondisi yang menghasilkan False: bukan_member_3007 akan bernilai False karena pelanggan diinput dengan tipe "member" (negasi dari True adalah False). Alternatif lain, jika kode promo yang dimasukkan pengguna adalah "KUPONPALSU", maka ekspresi keanggotaan promo_tersedia_3007 akan bernilai False.

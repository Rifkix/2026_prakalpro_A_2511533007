## 1.1 Latar Belakang
Perulangan (*looping*) merupakan salah satu struktur kontrol dasar dalam pemrograman yang memungkinkan sebuah blok kode dijalankan berulang kali tanpa harus menuliskan perintah yang sama berkali-kali. Tanpa perulangan, programmer harus menulis satu per satu perintah untuk setiap pengulangan, sehingga kode menjadi panjang, sulit dirawat, dan tidak fleksibel terhadap jumlah pengulangan yang ditentukan oleh pengguna. Dalam bahasa Python, perulangan diimplementasikan melalui *for* dan *while*. Perulangan *for* umumnya digunakan ketika jumlah pengulangan sudah diketahui dan sering dikombinasikan dengan fungsi *range()* untuk menghasilkan urutan bilangan.

Perulangan juga dapat disusun bersarang (*nested for*), yaitu sebuah perulangan yang berada di dalam perulangan lain. Pada struktur ini, perulangan dalam dijalankan sampai selesai untuk setiap satu kali iterasi perulangan luar, sehingga sangat cocok untuk membentuk data atau pola dua dimensi yang terdiri dari baris dan kolom, seperti persegi, segitiga, tabel penjumlahan, hingga pola bangun geometri berbingkai. Pada praktikum ini, pembentukan pola dilakukan dengan memanfaatkan parameter *end* pada fungsi *print()* serta percabangan *if* di dalam perulangan untuk menentukan karakter yang dicetak pada setiap posisi.

## 1.3 Manfaat

## 2.1 Dasar Teori
Dasar teori dalam praktikum ini meliputi enam konsep utama perulangan dalam Python:

1. **Perulangan For**: Struktur perulangan yang menjalankan blok kode (yang diberi indentasi) untuk setiap nilai pada suatu urutan. Bentuk umumnya adalah `for variabel in range(...):`. Variabel perulangan (misalnya `i`) akan berubah nilainya pada setiap iterasi dan dapat digunakan di dalam blok perulangan.

2. **Fungsi range()**: Menghasilkan urutan bilangan bulat untuk perulangan *for*. `range(stop)` menghasilkan bilangan dari 0 sampai stop-1, `range(start, stop)` menghasilkan bilangan dari start sampai stop-1, dan `range(start, stop, step)` menghasilkan bilangan dengan selisih sebesar step. Nilai *stop* tidak pernah ikut dihasilkan, dan *step* bernilai negatif (misalnya `range(n, 0, -1)`) digunakan untuk perulangan menurun dari n sampai 1.

3. **Parameter end pada print()**: Secara bawaan `print()` akan pindah baris setelah mencetak. Dengan `end=""` pencetakan berikutnya tetap berada pada baris yang sama tanpa jarak, sedangkan `end=" "` memberi satu spasi. Perintah `print()` tanpa argumen digunakan untuk pindah ke baris berikutnya.

4. **Variabel Akumulator**: Variabel yang diberi nilai awal (misalnya 0) lalu diperbarui pada setiap iterasi, misalnya `jumlah = jumlah + i`, sehingga di akhir perulangan variabel tersebut menyimpan hasil penjumlahan seluruh nilai.

5. **Nested For**: Perulangan *for* yang berada di dalam perulangan *for* lainnya. Perulangan dalam akan dijalankan penuh untuk setiap satu iterasi perulangan luar. Umumnya perulangan luar mengatur jumlah baris dan perulangan dalam mengatur jumlah kolom. Batas perulangan dalam dapat bergantung pada variabel perulangan luar, seperti pada pola segitiga.

6. **Percabangan di dalam Perulangan**: Struktur *if-elif-else* yang diletakkan di dalam perulangan digunakan untuk menentukan karakter apa yang dicetak pada suatu posisi baris dan kolom, misalnya membedakan karakter bingkai, isi pola, dan spasi.

### 2.2.3 Program "perulangan_for3_2511533007"

```python
ulang_3007 = int(input("Masukkan jumlah perulangan: "))

jumlah_3007 = 0
for i_3007 in range(1, ulang_3007 + 1):
    print(i_3007, end=" ")
    jumlah_3007 = jumlah_3007 + i_3007

    if i_3007 < ulang_3007:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_3007, end=" ")
print()
print("Jumlah =", jumlah_3007)
```

```python
tinggi_3007 = int(input("Masukkan tinggi pola (bilangan genap, misalnya 10): "))

if tinggi_3007 % 2 != 0:
    print("Tinggi harus bilangan genap")
else:
    a_3007 = tinggi_3007
    c_3007 = a_3007
    lebar_3007 = (2 * tinggi_3007) - 2

    for i_3007 in range(1, tinggi_3007 + 1):
        b_3007 = c_3007 + 1

        for j_3007 in range(1, lebar_3007 + 1):

            #Baris atas dan bawah
            if i_3007 == 1 or i_3007 == tinggi_3007:
                if j_3007 == 1 or j_3007 == lebar_3007:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_3007 == 1 or j_3007 == lebar_3007:
                    print("|", end="")
                else:
                    if j_3007 == c_3007:
                        print("<", end="")
                    elif j_3007 == b_3007:
                        print(">", end="")
                    elif j_3007 == (lebar_3007 - c_3007):
                        print("<", end="")
                    elif j_3007 == (lebar_3007 - c_3007 + 1):
                        print(">", end="")
                    elif j_3007 > b_3007 and j_3007 < (lebar_3007 - c_3007):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3007 -= 2

        if a_3007 <= 0:
            c_3007 = (-a_3007) + 2
        else:
            c_3007 = a_3007 
                    

```

3. Perulangan luar `i_3007` mengatur baris 1 sampai `tinggi_3007`, dan perulangan dalam `j_3007` mengatur kolom 1 sampai `lebar_3007`. Di dalam perulangan dalam, percabangan menentukan karakter pada setiap posisi: baris pertama dan terakhir berupa bingkai `#` pada kedua ujung dengan `=` di tengah; baris isi diberi `|` pada kolom pertama dan terakhir; kolom `c_3007` dan `lebar_3007 - c_3007` diberi `<`, kolom `b_3007` dan `lebar_3007 - c_3007 + 1` diberi `>`; kolom di antara pasangan tersebut diisi `.`; sisanya diisi spasi.

## Tugas Praktikum

Soal :

Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.

**Kode Jawaban :**

a. Kode Program "Tugas5_2511533007"

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3007 = int(input("Masukkan ukuran skala jam pasir (N): "))

if n_3007 <= 0:
    print("N harus bilangan bulat positif")
else:
    # Border atas: # + (4N + 5) karakter "=" + #
    print("#", end="")
    for i_3007 in range(4 * n_3007 + 5):
        print("=", end="")
    print("#", end="")
    print()

    # Fase 1: jam pasir atas (baris N turun s.d. 1)
    for baris_3007 in range(n_3007, 0, -1):
        # Garis tegak kiri dan satu spasi padding
        print("|", end="")
        print(" ", end="")

        # Spasi penyeimbang kiri: 2 * (N - baris)
        for spasi_3007 in range(2 * (n_3007 - baris_3007)):
            print(" ", end="")

        # Deret angka mundur: baris turun ke 1, dipisahkan spasi
        for angka_3007 in range(baris_3007, 0, -1):
            print(angka_3007, end=" ")

        # Poros kristal
        print("<*>", end="")

        # Deret angka maju: 1 naik ke baris, diawali spasi
        for angka_3007 in range(1, baris_3007 + 1):
            print(" ", end="")
            print(angka_3007, end="")

        # Spasi penyeimbang kanan: 2 * (N - baris)
        for spasi_3007 in range(2 * (n_3007 - baris_3007)):
            print(" ", end="")

        # Satu spasi padding dan garis tegak kanan
        print(" ", end="")
        print("|", end="")
        print()

    # Fase 2: poros titik pusat
    print("|", end="")
    for spasi_3007 in range(2 * n_3007 + 1):
        print(" ", end="")
    print("<*>", end="")
    for spasi_3007 in range(2 * n_3007 + 1):
        print(" ", end="")
    print("|", end="")
    print()

    # Fase 3: jam pasir bawah (baris 1 naik s.d. N)
    for baris_3007 in range(1, n_3007 + 1):
        print("|", end="")
        print(" ", end="")

        for spasi_3007 in range(2 * (n_3007 - baris_3007)):
            print(" ", end="")

        for angka_3007 in range(baris_3007, 0, -1):
            print(angka_3007, end=" ")

        print("<*>", end="")

        for angka_3007 in range(1, baris_3007 + 1):
            print(" ", end="")
            print(angka_3007, end="")

        for spasi_3007 in range(2 * (n_3007 - baris_3007)):
            print(" ", end="")

        print(" ", end="")
        print("|", end="")
        print()

    # Border bawah
    print("#", end="")
    for i_3007 in range(4 * n_3007 + 5):
        print("=", end="")
    print("#", end="")
    print()
```

**Penjelasan :**

Pertama, program mencetak judul dan meminta ukuran skala jam pasir `n_3007` yang dikonversi ke integer dengan `int()`. Percabangan `if n_3007 <= 0` memastikan N positif; jika tidak, program hanya mencetak pesan kesalahan, sedangkan seluruh proses pencetakan pola berada pada blok `else`.

Bingkai atas dicetak dengan `print("#", end="")`, lalu perulangan `for i_3007 in range(4 * n_3007 + 5)` mencetak karakter `=` satu per satu sebanyak 4N + 5 kali tanpa perkalian string, kemudian ditutup dengan `#` dan `print()` untuk pindah baris. Blok yang sama digunakan lagi di bagian akhir program sebagai bingkai bawah.

Fase 1 (jam pasir atas) menggunakan perulangan `for baris_3007 in range(n_3007, 0, -1)` sehingga `baris_3007` bergerak turun dari N sampai 1. Di dalamnya, program mencetak `|` dan satu spasi, lalu perulangan `range(2 * (n_3007 - baris_3007))` mencetak spasi penyeimbang kiri. Deret angka mundur dicetak oleh `for angka_3007 in range(baris_3007, 0, -1)` dengan `end=" "`, dilanjutkan poros `<*>`. Deret angka maju dicetak oleh `for angka_3007 in range(1, baris_3007 + 1)` yang mencetak satu spasi terlebih dahulu sebelum setiap angka. Setelah itu perulangan spasi penyeimbang kanan dengan jumlah yang sama seperti sisi kiri, satu spasi padding, `|`, dan `print()` untuk pindah baris. Semakin kecil nilai `baris_3007`, semakin banyak spasi penyeimbang sehingga bentuk menyempit menuju pusat.

Fase 2 (poros titik pusat) mencetak `|`, kemudian dua perulangan masing-masing sebanyak 2N + 1 spasi yang mengapit poros `<*>`, lalu `|`. Fase 3 (jam pasir bawah) memiliki isi baris yang sama dengan Fase 1, hanya saja perulangan barisnya menggunakan `range(1, n_3007 + 1)` sehingga `baris_3007` naik dari 1 sampai N dan bentuk kembali melebar.

Dengan susunan tersebut setiap baris pola memiliki panjang yang sama, yaitu 4N + 7 karakter (sama dengan lebar bingkai), sehingga garis tegak `|` pada sisi kiri dan kanan selalu sejajar. Program ini juga memenuhi ketentuan soal karena seluruh variabel berakhiran `_3007`, seluruh spasi, border, dan angka dibentuk dengan *for* bersarang berparameter `end=""`, dan tidak ada perkalian string.

# BAB III
# KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan *for* pada Python sangat efektif untuk menjalankan blok kode secara berulang dengan jumlah pengulangan yang jelas. Fungsi `range()` dengan satu, dua, atau tiga parameter (termasuk *step* negatif) memberikan kendali penuh atas nilai awal, batas, dan arah perulangan, sedangkan variabel akumulator memungkinkan perhitungan seperti penjumlahan deret dilakukan di dalam perulangan. Struktur *nested for* terbukti cocok untuk membentuk pola dua dimensi, di mana perulangan luar mengatur baris dan perulangan dalam mengatur kolom, dengan parameter `end=""` dan `print()` tanpa argumen untuk mengendalikan perpindahan baris. Penambahan percabangan *if* di dalam perulangan dalam memungkinkan setiap posisi pada pola menentukan karakternya sendiri, seperti pada pola berbingkai. Melalui studi kasus Jam Pasir Kristal Palindromik Berbingkai, mahasiswa dapat memahami cara menggabungkan *nested for*, `range()` menurun dan menaik, serta perhitungan jumlah spasi untuk menghasilkan pola yang simetris tanpa menggunakan perkalian string.

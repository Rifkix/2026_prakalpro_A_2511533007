# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
Dasar teori dalam praktikum ini meliputi beberapa konsep dasar pemrograman Python:

1. **Fungsi Output print()**: Digunakan untuk menampilkan pesan atau data ke layar. Pemanggilan print() tanpa argumen akan mencetak baris kosong.
2. **Komentar dalam Python**: Digunakan untuk memberi catatan pada kode tanpa dieksekusi oleh interpreter. Komentar satu baris menggunakan #, sedangkan komentar multi baris menggunakan triple quotes ("""...""").
3. **String Concatenation & Escape Character**: String dapat digabungkan menggunakan operator +. Karakter escape diawali tanda backslash (\\), seperti \n untuk baris baru (*newline*) dan \\\ untuk menampilkan karakter backslash.

## 2.2 Kode Program

### 2.2.1 Program "Program1_2511533007"

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/\\//\\\\///\\\\\\")
print("program ini" + " dibuat oleh a\n" + "rifki dan 2511533007.")
```

**Penjelasan :**
1. Baris # Program pertama dan komentar """...""" digunakan sebagai catatan/dokumentasi program yang diabaikan saat eksekusi.
2. print("Hello World") berfungsi menampilkan string "Hello World" ke layar.
3. print() kosong digunakan untuk mencetak satu baris kosong.
4. print("Program ini menghasilkan") dan print("Empat baris output") mencetak teks masing-masing baris.
5. print("/\\//\\\\///\\\\\\") mencetak kombinasi garis miring dan backslash. Setiap penulisan dua backslash (`\\`) menghasilkan satu karakter backslash (\\) pada tampilan output.
6. print("program ini" + " dibuat oleh a\n" + "rifki dan 2511533007.") menggabungkan tiga potongan string dengan operator +, serta menyisipkan karakter escape \n untuk membuat baris baru sebelum teks "rifki dan 2511533007."

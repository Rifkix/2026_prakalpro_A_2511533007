## 2.1 Dasar Teori
1. **Tipe Data Numerik**: Beberapa tipe data numerik yang sering digunakan di python yaitu int untuk angka bilangan bulat, float untuk bilangan berkoma, complex untuk bilangan kompleks
2. **Tipe Data String**: String merupakan sekumpulan karakter yang diapit oleh tanda petik tunggal ('), petik ganda ("), atau petik tiga (""") untuk string multiline.
3. **Tipe Data Boolean**: Boolean adalah tipe data yang hanya memiliki dua nilai, yaitu True atau False.
4. **Input dan Output**: Fungsi input() digunakan untuk menerima input dari pengguna berbentuk string. Untuk melakukan operasi perhitungan, input tersebut perlu diubah terlebih dahulu menggunakan fungsi seperti int() atau float(). Fungsi print() digunakan untuk menampilkan output ke layar.
5. **Konstanta**: Konstanta adalah variabel yang nilainya tidak boleh diubah sepanjang eksekusi program. Pada Python, konstanta menggunakan petunjuk tipe Final dari typing.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2511533007"


```python
is_lulus_3007 = True
is_cumlaude_3007 = False

#menggunakan Boolean
nilai_3007 = 85
batas_lulus_3007 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_3007 = nilai_3007 >= batas_lulus_3007 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3007)
print("Apakah Lulus?:", status_kelulusan_3007)
if is_lulus_3007 and is_cumlaude_3007:
    print("Selamat, Anda lulus dengan predikat cumlaude!")
```

**Penjelasan :**
1. Variabel `is_lulus_3007` bernilai `True` dan `is_cumlaude_3007` bernilai `False`.
2. `nilai_3007 >= batas_lulus_3007` membandingkan nilai `85` dengan `75`, yang menghasilkan nilai `True` pada `status_kelulusan_3007`.
3. Kondisi `if is_lulus_3007 and is_cumlaude_3007:` tidak menghasilkan apa apa karena `and` butuh kedua kondisi `True` sedangkan `is_cumlaude_3007` bernilai `False`

### 2.2.2 Program "Input_2511533007"

```python
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_3007 = input("Masukkan nilai panjang: ")
lebar_3007 = input("Masukkan nilai lebar: ")
luas_3007 = int(panjang_3007) * int(lebar_3007)
print("Luas =", luas_3007)
```

### 2.2.3 Program ("Konstanta_2511533007")

```python
from typing import Final

PI_3007: Final = 3.14
print("pi: %f" % (PI_3007))
jari_3007 = float(input("Masukkan nilai jari-jari: "))
luas_3007 = PI_3007 * jari_3007 * jari_3007
print(
    "Luas lingkaran dengan jari-jari %.2f adalah %.2f"
    % (jari_3007, luas_3007)
)
```

### 2.2.4 Program ("numerik1_2511533007")

```python
#create a variable with integer value
a_3007 = 100
print("The type of variable having value ", a_3007, "is", type(a_3007))

#create a variable with float value
b_3007 = 10.2345
print("The type of variable having value ", b_3007, "is", type(b_3007))

#create a variable with complex value
c_3007 = 100+3j
print("The type of variable having value ", c_3007, "is", type(c_3007))
```

**Penjelasan :**
1. Variabel `a_3007` berisi nilai integer `100`. Fungsi `type(a_3007)` memperlihatkan tipe datanya yaitu: `<class 'int'>`.
2. Variabel `b_3007` berisi nilai desimal `10.2345`. Fungsi `type(b_3007)` memperlihatkan tipe datanya yaitu: `<class 'float'>`.
3. Variabel `c_3007` berisi nilai bilangan kompleks `100+3j`. Fungsi `type(c_3007)` memperlihatkan tipe datanya yaitu: `<class 'complex'>`.

### 2.2.5 Program ("string_2511533007")

4. Variabel `agama_3007` menyimpan string `"islam"`. Seluruh variabel ditampilkan menggunakan `print()`.
